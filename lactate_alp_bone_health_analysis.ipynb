```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np

# Filter orthopaedic participants
ortho = df[df["Group"] == "orthopaedic"]

# Define X and Y
X = ortho["ALP (U/L)"].values.reshape(-1, 1)   # Independent variable
y = ortho["(mg/dL)"].values.reshape(-1, 1)  # Dependent variable

# Fit linear regression model
model = LinearRegression()
model.fit(X, y)

# Regression parameters
slope = model.coef_[0][0]
intercept = model.intercept_[0]
r_squared = model.score(X, y)

print("Regression equation: Lactate = {:.3f} * ALP + {:.3f}".format(slope, intercept))
print("R-squared:", r_squared)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression

# Filter orthopaedic participants
ortho = df[df["Group"] == "orthopaedic"]

# Define X (ALP) and Y (Lactate)
X = ortho["ALP (U/L)"].values.reshape(-1, 1)
y = ortho["(mg/dL)"].values.reshape(-1, 1)

# Fit linear regression model
model = LinearRegression()
model.fit(X, y)

# Predict Y for regression line
y_pred = model.predict(X)

# Get regression parameters
slope = model.coef_[0][0]
intercept = model.intercept_[0]
r_squared = model.score(X, y)

# Plot
plt.figure(figsize=(8,6))
plt.scatter(X, y, color='blue', label='Data points')
plt.plot(X, y_pred, color='red', linewidth=2, label='Regression line')

# Annotate regression equation and R²
plt.text(
    X.min(), 
    y.max(), 
    f"Lactate = {slope:.3f} * ALP + {intercept:.3f}\nR² = {r_squared:.3f}", 
    fontsize=12, 
    color='red'
)

plt.xlabel("ALP (U/L)")
plt.ylabel("(mg/dL)")
plt.title("Regression of Lactate on ALP (Orthopaedic Group)")
plt.legend()
plt.show()
```

```python
from scipy.stats import shapiro

# Put the names of your biomarkers/columns here
variables = ['ALP (U/L)', '(mg/dL)',]   # <-- change if needed

# Put the name of your group column here
group_column = 'Group'                 # <-- change if your column has another name

# Loop through each variable and each group
for var in variables:
    print(f"\nNormality test for {var}:")
    for grp in df[group_column].unique():
        subset = df[df[group_column] == grp][var].dropna()
        stat, p = shapiro(subset)
        print(f"  {grp}: W = {stat:.4f}, p = {p:.6f}")
```

```python
from scipy.stats import mannwhitneyu

# ALP comparison
ortho_alp = df[df['Group'] == 'orthopaedic']['ALP (U/L)']
non_alp = df[df['Group'] == 'non-orthopaedic']['ALP (U/L)']

u_stat, p_val = mannwhitneyu(ortho_alp, non_alp, alternative='two-sided')
print("ALP Mann–Whitney U:", u_stat, "p =", p_val)

# Lactate comparison
ortho_lac = df[df['Group'] == 'orthopaedic']['(mg/dL)']
non_lac = df[df['Group'] == 'non-orthopaedic']['(mg/dL)']

u_stat, p_val = mannwhitneyu(ortho_lac, non_lac, alternative='two-sided')
print("Lactate Mann–Whitney U:", u_stat, "p =", p_val)
```
